```python
x = data.drop(columns=['is_promoted'])
y = data['is_promoted']
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 42)
```

```python
education_map = {
    "Below Secondary": 0,
    "Bachelor's": 1,
    "Master's & above": 2
}
x_train['education'] = x_train['education'].map(education_map)
x_test['education'] = x_test['education'].map(education_map)
test_data['education'] = test_data['education'].map(education_map)

gender_map = {'m': 1, 'f': 0}
x_train['gender'] = x_train['gender'].map(gender_map)
x_test['gender'] = x_test['gender'].map(gender_map)
test_data['gender'] = test_data['gender'].map(gender_map)

nominal_cols = ['department', 'region', 'recruitment_channel']
x_train = pd.get_dummies(x_train, columns=nominal_cols, drop_first=True)
x_test = pd.get_dummies(x_test, columns=nominal_cols, drop_first=True)
test_data = pd.get_dummies(test_data, columns=nominal_cols, drop_first=True)
```

```python
x_train, x_test = x_train.align(x_test, join='left', axis=1, fill_value=0)
x_train, test_data = x_train.align(test_data, join='left', axis=1, fill_value=0)

print("x_train shape:", x_train.shape)
print("x_test shape: ", x_test.shape)
print("test_data shape:", test_data.shape)
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from sklearn.metrics import f1_score, accuracy_score, classification_report

models = {
    "Decision Tree": DecisionTreeClassifier(max_depth=10, random_state=42),
    "Random Forest": RandomForestClassifier(n_estimators=200, max_depth=12, random_state=42, n_jobs=-1),
    "XGBoost": XGBClassifier(n_estimators=200, learning_rate=0.08, max_depth=6, random_state=42, eval_metric='logloss')
}

best_model = None
best_model_name = ""
best_f1 = 0
best_acc = 0

print("=== MODEL COMPARISON EVALUATION ===\n")

for name, clf in models.items():
    clf.fit(x_train, y_train)
    preds = clf.predict(x_test)

    acc = accuracy_score(y_test, preds)
    f1 = f1_score(y_test, preds)

    print(f"--- {name} ---")
    print(f"Accuracy: {acc:.4f} | F1-Score: {f1:.4f}")

    if f1 > best_f1:
        best_f1 = f1
        best_acc = acc
        best_model_name = name
        best_model = clf

print("="*40)
print(f" BEST MODEL SELECTED: {best_model_name}")
print(f"Best Validation Accuracy: {best_acc:.4f}")
print(f"Best Validation F1-Score: {best_f1:.4f}")
print("="*40)

print(f"\nDetailed Classification Report for {best_model_name}:")
print(classification_report(y_test, best_model.predict(x_test)))
```

```python
from sklearn.model_selection import GridSearchCV, StratifiedKFold
from xgboost import XGBClassifier
from sklearn.metrics import f1_score, accuracy_score, classification_report

param_grid = {
    'max_depth': [4, 6, 8],
    'learning_rate': [0.03, 0.08, 0.1],
    'n_estimators': [150, 250],
    'subsample': [0.8, 1.0],
    'scale_pos_weight': [1, 2, 3]
}

base_xgb = XGBClassifier(random_state=42, eval_metric='logloss')

grid_search = GridSearchCV(
    estimator=base_xgb,
    param_grid=param_grid,
    scoring='f1',
    cv=StratifiedKFold(n_splits=5),
    verbose=1,
    n_jobs=-1
)

print("=== STARTING GRIDSEARCHCV FINE-TUNING ===")
grid_search.fit(x_train, y_train)

tuned_xgb = grid_search.best_estimator_
print("\n Best Hyperparameters Found:", grid_search.best_params_)

tuned_preds = tuned_xgb.predict(x_test)
tuned_acc = accuracy_score(y_test, tuned_preds)
tuned_f1 = f1_score(y_test, tuned_preds)

print("\n" + "="*40)
print(f"🏆 TUNED XGBOOST VALIDATION ACCURACY: {tuned_acc:.4f}")
print(f"🏆 TUNED XGBOOST VALIDATION F1-SCORE: {tuned_f1:.4f}")
print("="*40)

print("\nDetailed Classification Report for Fine-Tuned XGBoost:")
print(classification_report(y_test, tuned_preds))
```
